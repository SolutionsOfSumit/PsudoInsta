## Path Operation 

````python
@app.get('/')
def root():
    return {"message": "Hello World"}
````

This above code is called ***path operation*** which are also called routes in other programming languages but they fundamentally means the same thing.

If you see closely the whole operation is made up with two parts one is the decorator which is `@app.get('/)` and another is a function `async def root()`.

### async def root()
is a plain old python function accompanied with the ***async*** keyword.

What ***async*** does here?
It is optional you can start the function with `def` as well if you want to, `async` is helpful if you are writing a asynchronous code block which will take a certain amount of time.

***root*** over here is the name of our function which is given by the documentation itself. If can change it according to your requirement but try to keep the name as much descriptive as possible.

And In this function we are just returning a python dictionary.
***FastAPI will automatically convert this to JSON and send it to our webpage***

### @app.get('/')
This is a decorator which converts the plain old function to an actual path operator with out this we can not use fastAPI at all so that someone wants to use the api, they can hit this endpoint. @app is referring to the instance of the FastAPI which we have created. and .get() is one of the http methods. And inside the http method is the path ('/') which in above case is root, which means the path after your specific domain name of api.

# Schemas

We get to extract every timpe of information form the body and we don't know that which type of data will the user will send.
So we don't want the front end to send arbortary data and on top of that data is't geting velidated at all, for example what if the user sends a blank title and we don't have a post for blank title because we don't have anything to velidate it so we kind of force user to send a prticular type of data which is called schema.

***Schema*** is a defination of the information which our software will accept.

For tha we will use ***pydantic*** this was automatically installed when we have installed when we have installed it with FastAPI using [all] flag.

Although we have installed with FastAPI it is a stand alone lib of its own and it does not belong to FastAPI as its own method or function.


To import pydantic we will use 

````python
from pydantic import BaseModel
````

## How to create a schema 

For that we will create a class and extend it with `BaseModel` which we have imported form pydantic.

And in that we will pass in the different properties which we want or data with the field and the datatype that it will accept.

````python
class Post(BaseModel):
    title: str
    content: str
````

This above class will act like a schema and will force the user to use data in this structure. 
So now we can reference this class mode in our post request 

````python
@app.post("/create_post")
def create_post(new_post: Post):
    print(new_post)
    return {"data": "new_data"}
````

You can ask for specific data as well 

````python
@app.post("/")
def create_post(new_post: Post):
    print(new_post.content)
    return {"data": "Operation successful"}
````

If the integer is provided instead of a string, it will be automatically will be get converted into string as well.

### Default values in schema 

If you want to make some values default that if user will not provide then to it will auto assign themselves.


````python
class Posst(BaseModel):
    title: str
    content: str
    make_live: bool = True # this is a Default value
````

this does not only work with `bool` it can also work with a particular kind of string and integer.

### Optional values 
These are the values if the user will not provide anything they will remain none and will not store a default value.

Now to create an Optional vale we have to import Optional form typing field and provide the datatype to be stored in it?

````python
from typing import Optional

class Post(BaseModel):
    title: str
    content: str
    rating: Optional[int] = None 
````

So by using this pydantic models we can ensure the validity of the data which is send by the user to our software.

````python
@app.post("/create_post")
def create_post(new_post: Post):
    print(new_post)
    return {"data": "new_data"}
````

The above code resembles that we are storing the data in a variable called `new_post` there for the data is getting stored in a separate pydantic model.

Now if you ever need to convert your data which is in pydantic model to a dictionary you can add `.dict` after the model

````python
@app.post("/create_post")
def create_post(new_post: Post):
    print(new_post.model_dump())
    return {"data": "new_data"}
````

but in pydantic V2 `.dict` is replaced with `.model_dump` 

# CRUD applications 
CRUD is a n acronym which stands for 3 main functions of the application.
And there are some standard convention for naming APIs.

Like say if your are working with the program which work with posts its likely that you will name all your URLs as `/posts` not `/post`which is that 


1. You have to write the path name as plural.
```python
@app.post('/posts')
```


2. Any time you want to create an entity its gonna be a `post` request.
3. For reading both an individual or multiple post we use `get` request.
```python 
@app.get('/posts')
```
 for all the posts and 
```python 
@app.get('/posts/{id}')
```
 for a specific post.
4. For updating we use two types of request `put` and `patch`.
    And the difference b/w both of them is for updating through you have to pass every filed required in the schema and for patch you just have to pass the field which have to be updated.
```python 
@app.put('/posts/{id}')
```

5. And the last one is delete which is simple and states by its name that it is used to delete the contents.
```python 
@app.delete('/posts/{id})
```
