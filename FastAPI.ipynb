### How to retrieve a single post

For that we have to use id which we have created earlier and the decorator will look something like this
```python 
@app.get('/posts/{id}')
```
and here `{id}` will be provided by the user and it is also called path parameter.
The amazing thing is FastAPI automatically extracts it and then we can directly provide it in our function.

````python
@app.get('/posts/{id}')
def get_post2(id):
    return {"data": f"Here is your{id}"}
````

The above code shows how FastAPI gets data form url now we have to write the logic that will search for the specific item in the memory using the received parameter.

***But the id which was provided by the parameter will be in the string format so before we can use it we have to convert it in favour of our project.***

````python
@app.get('/posts/{id}')
def get_post2(id):
    for post in my_memory:
        if id == my_memory['id']:
            return {"data": post}
````

How ever the best practice is to write function separately and then call it inside our path operator. 

````python
def find_post(id):
    for post in my_memory:
        if post["id"] == int(id):
            return post

@app.get('/posts/{id}')
def get_post(id):
    post = find_post(id)
    return {"data": post}
````

### Validation of path parameter 

In above code we have forced our code to only accept integer and if something else is given it will brake the server, for tackling that we can validate the path parameter by which FastAPI will first check if the parameter can be converted to the required data type and if not it will automatically send a response to the frontend without breaking the server.

````python
@app.get('/posts/{id}')
def get_post(id: int):     # this will validate and convert itself there is no need to do it elsewhere in code.
    post = find_post(id)
    return {"data": post}
````

### Position of the api matters 

for FastAPI paths like `/posts/{id}` is same as `/post/latest`

for example if your code is written like this

````python

@app.get('/posts/{id}')
def get_post(id: int):
    for post in my_memory:
        if post['id'] == id:
            return {"data": post}

@app.get('/posts/latest')
def get_latest():
    post =my_memory[ len(my_memory) - 1 ] 
    return {"data": post}
````

The FastAPI will go form top down one by one and come to the endpoint '/post/' and consider {id} == latest because {id} is a variable and it can be anything which means it can be latest as well.

And to fix this the prefer way is to write  '/post/latest' before '/post/{id}' because our values we will provide form {id} will never be == latest.

***It is preferred to write static routes before dynamic routs***

## Deleting request
for now because we are not using any database we will will write an function which will grab the index of the post and delete it.

````python
def find_post_index(id):
    for i, p in enumerate(my_memory):
        if p["id"] == id:
            return i
        
@app.delete('/post/{id}')
def delete_post(id):
    index = find_post_index(id)
    my_memory.pop(index)
    return {"message": "Your post is deleted"}
````

This will work but the status code for deleting anything will will be 204 which says no content but the issue with that is that it will not return any message back, so to resolve this we will write out code like below.

````python
@app.delete('/posts/{id}' status_code=status.HTTP_204_NO_CONTENT)
def delete_post(id: int):
    index = find_post_index(id)
    my_memory.pop(index)
    return Response(status_code=status.HTTP_204_NO_CONTENT)
````

if the user sends a request to delete the post which is not there then the serer will throw an error.

Well we can just add add an if statement before finding the index handling the exception.

````python
@app.delete('/posts/{id}')
def delete_post(id: int):
    if id == None:
        raise HTTPException(status_code=status.HTTP_404_NOT_FOUND, detail=f"The post with id {id}, no longer exist")
````

### Building APIs
for put request

````python
@app.put('/posts/{id}')
def update_post(id: int, post: Post):
    index = find_post_index(id)
    if index == None:
        raise HTTPException(status_code=status.HTTP_404_NOT_FOUND, detail="Post does not exist")
    post = post.model_dump()
    my_memory[index] = post
    return {"detail": post}
````

now `patch` request is little complicated here because in our pydantic class we have make two fields as necessary and if we will use that same class in our `patch` update even though we only have to update title we have to give the content as well.

To fix this we can either make a new pydantic model only for patch or we can make those fields optional 

````python
from typing import Optional

class Post(BaseModel):
    title: Optional[str] = None
    content: Optional[str] = None
````

Now we can use it for partial update as well

````python
@app.patch('/posts/{id}')
def partial_update(id: int, post: Post):
    index = find_post_index(id)
    if index == None:
        raise HTTPException(status_code=status.HTTP_404_NOT_FOUND, detail="Post Does not exist")
    post.model_dump(exclude_unset=True)      # This will only use the fields that user have provide and will make all the other fields null
    my_memory[index].update(post)            # This will only take the fields which are not null and update them with the fields in the entity
    print("Update successfully")
    return {"detail": my_memory[index]}
````

if we will not use `exclude_unset=True` that will make all other fields as null including id if we are not providing it later and will give use response some thing like this

````python
{
    "detail": {
        "title": "updagted title",
        "content": null,
        "id": 1,
        "publish": null,
        "rating": null
    }
}
````

where you can see the problem, the model was picked up by the pydantic model so it had all the other fileds but they all are null now, and if we use `exclude_unset=True` it will return us

````python
{
    "detail": {
        "title": "updated title",
        "content": "Contents of post 1",
        "id": 1
    }
}
````

As you can see the other fields are kept intect, but here one more thing is which is very crucial which is `.update`.

if we simply use `my_memory[index] = post` which at this point is just 

````python
{
	"title": "updated title"
}
````

By using the `.update` it will update the dictionary keeping other fields unlike replacing them entairly

## Fatching individual post by {id}

How ever there are many porblems in this and the question can be why we are using fetchone() and

Because fetchall() will continuesly search for the id again and again even though  we have found it.  

#### Another error will be
that the complete query we are passing will be as string but the id we are passing will be an integer which will cause issues in concatenation, to fix this we also have to send our id in string format.

***But now you can think we can do that by just changing the validataion*** 

````python
@app.get('/posts/{id}')
def get_post_by_id(id: str):
    cursor.execute("""SELECT * FROM posts WHERE id = %s""", (id))
    post = cursor.fetchone()
    return post
````

````python
@app.post('/posts/{id}')
def get_post_by_id(id: int):
    cursor.execute("""SELECT * FROM posts WHERE id = %s""", (str(id)))
    post = cursor.fetchone()
    return post
````
