## Path Operation 

````python
@app.get('/')
def root():
    return {"message": "Hello World"}
````

This above code is called ***path operation*** which are also called routes in other programming languages but they fundamentally means the same thing.

If you see closely the whole operation is made up with two parts one is the decorator which is `@app.get('/)` and another is a function `async def root()`.

### async def root()
is a plain old python function accompanied with the ***async*** keyword.

What ***async*** does here?
It is optional you can start the function with `def` as well if you want to, `async` is helpful if you are writing a asynchronous code block which will take a certain amount of time.

***root*** over here is the name of our function which is given by the documentation itself. If can change it according to your requirement but try to keep the name as much descriptive as possible.

And In this function we are just returning a python dictionary.
***FastAPI will automatically convert this to JSON and send it to our webpage***

### @app.get('/')
This is a decorator which converts the plain old function to an actual path operator with out this we can not use fastAPI at all so that someone wants to use the api, they can hit this endpoint. @app is referring to the instance of the FastAPI which we have created. and .get() is one of the http methods. And inside the http method is the path ('/') which in above case is root, which means the path after your specific domain name of api.

## How to create a schema 

For that we will create a class and extend it with `BaseModel` which we have imported form pydantic.

And in that we will pass in the different properties which we want or data with the field and the datatype that it will accept.

````python
class Post(BaseModel):
    title: str
    content: str
````

This above class will act like a schema and will force the user to use data in this structure. 
So now we can reference this class mode in our post request 

````python
@app.post("/create_post")
def create_post(new_post: Post):
    print(new_post)
    return {"data": "new_data"}
````

You can ask for specific data as well 

````python
@app.post("/")
def create_post(new_post: Post):
    print(new_post.content)
    return {"data": "Operation successful"}
````

If the integer is provided instead of a string, it will be automatically will be get converted into string as well.

### Default values in schema

### Optional values 
These are the values if the user will not provide anything they will remain none and will not store a default value.

Now to create an Optional vale we have to import Optional form typing field and provide the datatype to be stored in it?

````python
from typing import Optional

class Post(BaseModel):
    title: str
    content: str
    rating: Optional[int] = None 
````

So by using this pydantic models we can ensure the validity of the data which is send by the user to our software.

````python
@app.post("/create_post")
def create_post(new_post: Post):
    print(new_post)
    return {"data": "new_data"}
````

The above code resembles that we are storing the data in a variable called `new_post` there for the data is getting stored in a separate pydantic model.

Now if you ever need to convert your data which is in pydantic model to a dictionary you can add `.dict` after the model

````python
@app.post("/create_post")
def create_post(new_post: Post):
    print(new_post.model_dump())
    return {"data": "new_data"}
````

### Example of POST

For now we are not using a prodper database technology so for now we will store our data inside the memory and we can do that by storing everything inside a variable.

````python
class Post(BaseModel):
    title: str
    content: str
    id : int     
````

Every time we will use a database it will assign a id to the content which we upload or update. Here because we are not using one we have created a field of id for that.

````python
my_memory = [
    {
        "title": "Title of the 1 post",
        "content": "content of the 1 post",
        "id": 1
    },
        {
        "title": "Title of the 2 post",
        "content": "content of the 2 post",
        "id": 1
    }
    ]
````

This above list will work like an memory for us.

### updating memory form post req

because we are not using a proper data base we need to have random id assigned to our posts. For that we can use `randrange` function form `random` lib.

````python
from random import randrange
````

Now first we will convert our pydantic model to a dictionary and then append a random no. as an id inside the dictionary. 

````python
@app.post('/posts')
def create_post(post: Post):
    post_dict = post.model_dump()
    post_dict["id"] = randrange(0, 1000000000000000)
    my_memory.append(post_dict)
    return {"data": post_dict}
````

### How to retrieve a single post

For that we have to use id which we have created earlier and the decorator will look something like this
```python 
@app.get('/posts/{id}')
```
and here `{id}` will be provided by the user and it is also called path parameter.
The amazing thing is FastAPI automatically extracts it and then we can directly provide it in our function.

````python
@app.get('/posts/{id}')
def get_post2(id):
    return {"data": f"Here is your{id}"}
````

The above code shows how FastAPI gets data form url now we have to write the logic that will search for the specific item in the memory using the received parameter.

***But the id which was provided by the parameter will be in the string format so before we can use it we have to convert it in favour of our project.***

````python
@app.get('/posts/{id}')
def get_post2(id):
    for post in my_memory:
        if id == my_memory['id']:
            return {"data": post}
````

How ever the best practice is to write function separately and then call it inside our path operator. 

````python
def find_post(id):
    for post in my_memory:
        if post["id"] == int(id):
            return post

@app.get('/posts/{id}')
def get_post(id):
    post = find_post(id)
    return {"data": post}
````

### Validation of path parameter 

In above code we have forced our code to only accept integer and if something else is given it will brake the server, for tackling that we can validate the path parameter by which FastAPI will first check if the parameter can be converted to the required data type and if not it will automatically send a response to the frontend without breaking the server.

````python
@app.get('/posts/{id}')
def get_post(id: int):     # this will validate and convert itself there is no need to do it elsewhere in code.
    post = find_post(id)
    return {"data": post}
````

### Position of the api matters 

for FastAPI paths like `/posts/{id}` is same as `/post/latest`

for example if your code is written like this

````python

@app.get('/posts/{id}')
def get_post(id: int):
    for post in my_memory:
        if post['id'] == id:
            return {"data": post}

@app.get('/posts/latest')
def get_latest():
    post =my_memory[ len(my_memory) - 1 ] 
    return {"data": post}
````

The FastAPI will go form top down one by one and come to the endpoint '/post/' and consider {id} == latest because {id} is a variable and it can be anything which means it can be latest as well.

And to fix this the prefer way is to write  '/post/latest' before '/post/{id}' because our values we will provide form {id} will never be == latest.

***It is preferred to write static routes before dynamic routs***

## Changing exception status code

````python
from fastapi import Response
````

and then you can pass it in as a parameter in your operation function, with and variable  

````python
@app.get('/posts/{id}')
def get_post(id: int, response: Response):
    post = find_post(id)                   # If post not found this throws null. 
    if not post:                           # which validates this statement. 
        response.status_code = 404         # Hence running this statement.
    return {"data": post}
````

now there is a better way to do it.

In the above method we have to remember all the status codes which is hard code all the status code which will not be practical so we import them for our reference.

````python
from fastapi import Response, status
````

now in the path opertaiton instead hardcoding the status code we will take referance form `status`.

````python
@app.get('/posts/{id}')
def get_post(id: int, response: Response):
    post = find_post(id)
    if not post:
        response.status_code = status.HTTP_404_NOT_FOUND
    print("Post Found")
    return {"post": post}
````

There is one more thing we can do to help frontend which is, instead of automatically returning null we can write a return statement in this if block returning the error which we have got

````python
@app.get('/post/{id}')
def get_post(id: int, response: Response):
    post = find_post(id)
    if not post:
        response.status_code = status.HTTP_404_NOT_FOUND
        return {"message": f"post with id:{id} does not exist"}
    return {"post": post}
````
